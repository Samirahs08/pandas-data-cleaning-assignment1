Step 3: Handling Missing Values

```python
#To identify missing values in the dataset:

df.isnull().sum()
```

# After checking for missing values, I found that several columns contain null values.

- discount has 121 missing values.
- what_customers_said has 573 missing values.
- seller_name has 301 missing values.
- videos has 781 missing values.
- seller_information has 301 missing values.
- variations has 562 missing values.

Among these, videos and what_customers_said contain a large number of missing values.

Since columns such as videos and variations contain a large number of missing values and are not essential for this analysis, they were left unchanged.

For useful columns like discount, seller_name, and seller_information, missing values were filled appropriately.

```python
# Fill missing discount values with the average discount

df["discount"] = df["discount"].fillna(df["discount"].mean())

# Fill missing seller names

df["seller_name"] = df["seller_name"].fillna("Unknown")

# Fill missing seller information

df["seller_information"] = df["seller_information"].fillna("Not Available")
```

```python
#verify whether the above modified columns have modified or not

df[["discount","seller_name","seller_information"]].isnull().sum()
```

# Step 5: Removing Duplicate Records


To ensure data quality, I checked whether the dataset contained any duplicate records. Duplicate rows can lead to incorrect analysis and should be removed before further processing.

```python
#First, I checked whether duplicate records existed.

print(df.duplicated().sum())
```

# Observation:

The output shows that there are no duplicate records present in the dataset.

```python
#Then I removed them.

df = df.drop_duplicates()
```

```python
#Verification:

print(df.duplicated().sum())
```

Verification:

# Step 6: Creating a Derived Column

The assignment requires creating a new column called **total_amount** using the formula:

total_amount = price × quantity

Before performing this calculation, the **final_price** column needs to be cleaned because the values are stored as text and contain special characters such as quotation marks and commas. These values must be converted into a numeric format so that mathematical operations can be performed.

```python
# Clean the final_price column

df["final_price"] = (
    df["final_price"]
    .astype(str)
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.replace('"', "", regex=False)
    .str.strip()
)

# Convert to numeric datatype

df["final_price"] = pd.to_numeric(df["final_price"], errors="coerce")
```

# Creating Quantity and Total Amount Columns

The dataset does not contain a quantity column. For demonstration purposes, a quantity value of 1 is assigned to each product. Using this quantity value and the cleaned price column, a new column called total_amount is created.

```python
# Create quantity column

df["quantity"] = 1

# Create total_amount column

df["total_amount"] = df["final_price"] * df["quantity"]
```

```python
# Display the newly created columns

df[["final_price", "quantity", "total_amount"]].head()
```

# Observation:

The final_price column was successfully converted into a numeric format. A new column named total_amount was created using the formula:

total_amount = final_price × quantity

This derived column can be used for further analysis and calculations.
